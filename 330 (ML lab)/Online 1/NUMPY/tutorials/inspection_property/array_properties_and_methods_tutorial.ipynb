# NumPy Array Properties & Methods — Explanations

This cell explains each property and method used in the code below:

---

- **astype(dtype):** Converts all elements of the array to the specified data type (e.g., int, float, str). Useful for changing how data is stored or processed.
    - Example: `arr.astype(int)` turns all values into integers.
- **tolist():** Converts the NumPy array into a regular Python list. Useful for exporting data or using Python list methods.
    - Example: `arr.tolist()` gives a nested list for 2D arrays.
- **np.info(function):** Prints documentation about a NumPy function or object. Use this to learn what a function does and its parameters.
    - Example: `np.info(np.eye)` shows info about the identity matrix function.
- **size:** Returns the total number of elements in the array (all rows × columns × ...).
    - Example: `arr.size` for a 2x3 array returns 6.
- **shape:** Returns the dimensions of the array as a tuple (rows, columns, ...). Helps you understand the structure of your data.
    - Example: `arr.shape` for a 2x3 array returns (2, 3).
- **dtype:** Returns the data type of the array's elements (e.g., int64, float64). Useful for checking or debugging data types.
    - Example: `arr.dtype` might return `float64` if the array contains decimal numbers.

---

Run the code cell below to see how each property/method works in practice!

```python
import numpy as np

arr = np.array([[1.2, 2.5, 3.1], [4.0, 5.7, 6.3]])

# ASTYPE: Convert to integer type
arr_int = arr.astype(int)
print('astype(int):', arr_int)
print( 'original array : ',arr)

# TOLIST: Convert to Python list
arr_list = arr.tolist()
print('tolist():', arr_list)

# INFO: View documentation for np.eye
print('np.info(np.eye):')
np.info(np.eye)

# SIZE: Number of elements
print('size:', arr.size)

# SHAPE: Dimensions
print('shape:', arr.shape)

# DTYPE: Data type
print('dtype:', arr.dtype)
```
