# NumPy — Importing & Saving Tabular Data

This cell explains the most important parameters for `np.loadtxt`, `np.genfromtxt`, and `np.savetxt` in simple language, with examples for each option.

---

## Common Parameters Explained

### `delimiter`
- **What it does:** Sets the character that separates columns in your file.
- **Example:**
  - `delimiter=','` for CSV files (comma-separated)
  - `delimiter=' '` or leave blank for whitespace

### `dtype`
- **What it does:** Sets the data type for the loaded array (e.g., `float`, `int`, `str`).
- **Example:**
  - `dtype=float` (default for numbers)
  - `dtype=int` (for integer data)

### `skiprows` / `skip_header`
- **What it does:** Skips the first N lines of the file (useful if your file has a header row).
- **Example:**
  - `skiprows=1` (skip the first line)
  - `skip_header=2` (skip the first two lines)

### `usecols`
- **What it does:** Loads only specific columns from the file.
- **Example:**
  - `usecols=(0, 2)` loads only the first and third columns

### `converters`
- **What it does:** Lets you apply a function to a column as it is loaded (for custom parsing).
- **Example:**
  - `converters={0: lambda s: float(s) * 10}` multiplies the first column by 10 as it loads

### `missing_values` / `filling_values` (genfromtxt only)
- **What it does:** Tells NumPy what to treat as missing, and what value to use instead.
- **Example:**
  - `missing_values='NA', filling_values=0` replaces 'NA' with 0

### `fmt` (savetxt only)
- **What it does:** Sets the format for saving numbers (e.g., how many decimals).
- **Example:**
  - `fmt='%.2f'` saves numbers with 2 decimal places

## Quick Reference Table
| Parameter         | Used in         | What it controls                        | Example                      |
|-------------------|-----------------|-----------------------------------------|------------------------------|
| delimiter         | loadtxt, genfromtxt, savetxt | Column separator           | delimiter=','                |
| dtype             | loadtxt, genfromtxt          | Data type                  | dtype=float                  |
| skiprows/skip_header | loadtxt/genfromtxt      | Skip header lines           | skiprows=1                   |
| usecols           | loadtxt, genfromtxt          | Select columns             | usecols=(0,2)                |
| converters        | loadtxt, genfromtxt          | Custom column parsing      | converters={0: func}         |
| missing_values    | genfromtxt                   | What counts as missing     | missing_values='NA'          |
| filling_values    | genfromtxt                   | Value for missing entries  | filling_values=0             |
| fmt               | savetxt                      | Output number format       | fmt='%.2f'                   |

```python
# Example: create a small array, save it, and read it back using loadtxt/genfromtxt
import numpy as np
from tempfile import NamedTemporaryFile

# create arrays
arr = np.array([[1.5, 2.0, 3.25], [4.0, 5.75, 6.125]])
arr_mixed = np.array([[1, 2.0, 3], [4, np.nan, 6]])

# save to a temporary CSV file and read with loadtxt
with NamedTemporaryFile('w+', delete=False, suffix='.csv') as f:
    np.savetxt(f.name, arr, delimiter=',', fmt='%.3f', header='c1,c2,c3', comments='')
    path = f.name

print('Saved to', path)
print('loadtxt ->')
print(np.loadtxt(path, delimiter=','))

# save an array with a missing value and read with genfromtxt
with NamedTemporaryFile('w+', delete=False, suffix='.csv') as f2:
    np.savetxt(f2.name, arr_mixed, delimiter=',', fmt='%s')
    path2 = f2.name

print('Saved mixed to', path2)
print('genfromtxt -> (missing values converted to 0.0)')
print(np.genfromtxt(path2, delimiter=',', dtype=float, filling_values=0.0))
```
