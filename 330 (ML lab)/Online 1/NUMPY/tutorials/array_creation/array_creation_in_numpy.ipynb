# NumPy — Creating Arrays

This notebook explains the most common ways to create arrays in NumPy, as shown in the reference image. Each function is described with its syntax, a code example, and a plain English explanation.

---

## Common Array Creation Functions

| Function         | Example Code                                 | What it does                                      |
|------------------|----------------------------------------------|---------------------------------------------------|
| `np.array`       | `np.array([1, 2, 3])`                        | Create a 1D array from a Python list              |
|                  | `np.array([[1, 2, 3], [4, 5, 6]])`           | Create a 2D array from a list of lists            |
| `np.zeros`       | `np.zeros(3)`                                | 1D array of length 3, all values set to 0         |
| `np.ones`        | `np.ones((3, 4))`                            | 3x4 array, all values set to 1                    |
| `np.eye`         | `np.eye(5)`                                  | 5x5 identity matrix (1s on diagonal, 0 elsewhere) |

---

## Function Details

### `np.array`
- **Purpose:** Convert a Python list (or list of lists) into a NumPy array.
- **1D Example:**
  ```python
  arr = np.array([1, 2, 3])
  # arr: array([1, 2, 3])
  ```
- **2D Example:**
  ```python
  arr2d = np.array([[1, 2, 3], [4, 5, 6]])
  # arr2d: array([[1, 2, 3], [4, 5, 6]])
  ```

### `np.zeros`
- **Purpose:** Create an array filled with zeros.
- **Example:**
  ```python
  arr = np.zeros(3)
  # arr: array([0., 0., 0.])
  ```

### `np.ones`
- **Purpose:** Create an array filled with ones.
- **Example:**
  ```python
  arr = np.ones((3, 4))
  # arr: 3x4 array of ones
  ```

### `np.eye`
- **Purpose:** Create an identity matrix (ones on the diagonal, zeros elsewhere).
- **Example:**
  ```python
  arr = np.eye(5)
  # arr: 5x5 identity matrix
  ```

---


````python
# Import numpy
import numpy as np

# 1. Create a 1D array
arr1d = np.array([1, 2, 3])
print('1D array:', arr1d)

# 2. Create a 2D array
arr2d = np.array([[1, 2, 3], [4, 5, 6]])
print('2D array:\n', arr2d)

# 3. Array of zeros
zeros_arr = np.zeros(3)
print('Zeros array:', zeros_arr)

# 4. Array of ones
ones_arr = np.ones((3, 4))
print('Ones array:\n', ones_arr)

# 5. Identity matrix
eye_arr = np.eye(5)
print('Identity matrix:\n', eye_arr)
````
