```python
import numpy as np
arr= np.array([23,4,2,132])
new_arr=np.append(arr,[2,3])
print(f"old array {arr}")
print(f"new array {new_arr}")
```

```python
arr_inserted=np.insert(arr,1,3)
arr_inserted
```

```python
arr_2d=np.array([
    [1,2,3],
    [30,20,10],
    [3,4,5]
])

arr_row_deleted=np.delete(arr_2d,0,axis=0)
arr_col_deleted=np.delete(arr_2d,1,axis=1)
print(f"row deleted {arr_row_deleted}")
print(f"col deleted {arr_col_deleted}")
print(f"arr_2d (original) {arr_2d}")
```
