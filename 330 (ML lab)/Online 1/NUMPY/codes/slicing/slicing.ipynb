```python
import numpy as np
arr=np.array([2,3,4,5,20])
print(arr[0])
print(arr[1])
print(arr[-1])
print(arr[1:4])
print(arr[:4])
print(arr[::2])
```

```python
arr_2d=np.array([
    [1,2,3],
    [3,4,5],
    [9,8,7]    
])

print(arr_2d[1,2])
print(arr_2d[0:2, 0:2])
print(arr_2d[1:,1:])
print(arr_2d[0,:])
```

```python
# boolean masking and filtering 
mask= arr %2 ==0
print(mask)
print(arr)
print(arr[mask])
print(arr[arr%2==0])
```
