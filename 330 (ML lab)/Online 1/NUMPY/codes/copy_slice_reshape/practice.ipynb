```python
import numpy as np
arr= np.array([3.4,4.5,5,6,7])
new_arr=np.copy(arr)
# proving that changing old array deosnt change new array
# we actually create a new copy in memory and not just a new reference 

arr[2]=4543.3
print(arr)
print(new_arr)
```

```python
print(arr.dtype)
# creating a view with a diff daata type
arr_view= arr.view(np.int64)
print(arr)
print(arr_view)
# vochkaya gelo shobkisu.karon python jemne memory interpret korto shob cng hoye gese 
```

```python
# now we will see how to sort the array 
unsorted_array=np.array([3,4,5,6,1,2,3,4,0,32])
unsorted_array.sort()
print(unsorted_array)
```

```python
# now we will see how to sort rowwise or columnwise 

arr_2d=np.array([
    [3,5,1],
    [3,5,2],
    [90,1,100]
])
arr_2d_=np.copy(arr_2d)

arr_2d.sort(axis=0) 
# sorts row wise 

arr_2d_.sort(axis=1)

print(arr_2d)
print(arr_2d_)
```

```python
# flattening arrays 
arr_2d=np.array([
    [1,2,4],
    [4,5,6]
])
flat=arr_2d.flatten()
flat
```

```python
# reshaping array 
arr=np.arange(12)
print(f'before reshaping {arr}')
reshaped_arr= arr.reshape(3,4)
print(f'after reshaping {reshaped_arr}')
```
