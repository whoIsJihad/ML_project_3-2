```python
inventory = pd.DataFrame({
    'Item': ['Apple', 'Banana', 'Orange'],
    'Quantity': [50, 100, 75],
    'Price': [0.5, 0.3, 0.8]
})

print(inventory['Quantity'])
```

```python
print(inventory.iloc[2])
print(inventory.loc[1,'Price'])
print(inventory[inventory['Price']<=0.5])
```

```python
inventory['IsExpensive']= inventory['Price']>0.6
inventory['TotalSum']= inventory['Price']*inventory['Quantity']
inventory = inventory.rename(columns={'TotalSum':'Total'})
```
